# Step-2 Removing HTML Tags

```python
import re
def remove_html_tag(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r"" ,text)
```

```python
text = "<html><body> sddsdsffvfvsd <p> <p>"
```

```python
remove_html_tag(text)
```

```python
df['review'] = df['review'].apply(remove_html_tag)
```

```python
df
```

# Step-3 remove URL

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'' , text)
```

```python
text2 = "For Notebook go to https://www.github.com/Amaan-Syed-2182"
```

```python
remove_url(text2)
```

# Step-4 Remove Punctuation

```python
import string
string.punctuation


```

```python
exclude = string. punctuation

```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace (char , '')
    return text

```

```python

text1 = 'string. With. Punctuation?'

remove_punc(text1) 
```

```python
df['review'] = df['review'].apply(remove_punc)
```

```python
df
```

# Step-5 Chatword Treatment

```python
chat_words = {'u2' : 'you too' , 'BBL': 'Be Back Later' , 'GAL': 'Get A Life'}

chat_words

{'u2': 'you too', 'BBL': 'Be Back Later', 'GAL': 'Get A Life'}

```

```python
def chat_conversation(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words :
            new_text.append(chat_words[w.upper()])
        else :
            new_text.append(w)

    return "".join(new_text)
```

```python

chat_conversation('BBL')
```
